```python
## Department Escalation Analysis
#Which departments have the highest escalation levels?
department_escalations = df.groupby("department")["escalations"].agg([ "sum" , "mean" , "count"])
department_escalations
```

```python
df.groupby("department")["recurring_escalations"].agg(
    ["sum", "mean", "count"]
)
```

```python
department_escalation_rate = (
    df.groupby("department")["escalations"]
    .apply(lambda x: (x > 0).mean() * 100))

department_escalation_rate
```

```python
department_escalation_rate.sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.xlabel("Employees with at Least One Escalation (%)")
plt.ylabel("Department")
plt.title("Employee Escalation Rate by Department")
plt.show()
```

```python
reason_analysis = df.groupby("escalation_reason")["escalations"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

reason_analysis
```

```python
#How many employees are associated with each escalation reason?
df["escalation_reason"].value_counts()
```

```python
df["severity"].value_counts()
```

```python
#High/Critical escalation analysis
high_severity = df[df["severity"].isin(["High", "Critical"])]

high_severity.groupby("department")["escalations"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)
```

```python
#Are employees with more customer interactions experiencing more escalations?
plt.figure(figsize=(8, 6))

plt.scatter(
    df["customer_interactions"],
    df["escalations"],
    alpha=0.5
)

plt.xlabel("Customer Interactions")
plt.ylabel("Number of Escalations")
plt.title("Customer Interactions vs Escalations")

plt.show()
```

```python
df["customer_interactions"].corr(df["escalations"])
#Customer interactions showed a moderate positive correlation with escalation count (r = 0.555), 
#suggesting that higher interaction workloads are associated with more escalations.
# However, correlation does not establish causation, and other operational factors may also contribute.
```

```python
##Do employees working more overtime tend to have more escalations?
df["overtime_hours"].corr(df["escalations"])
#Customer interactions showed a stronger positive correlation with escalations (r = 0.555) 
# than overtime hours (r = 0.2533). This suggests workload volume may have a stronger association with escalations 
# than overtime alone, although these correlations do not establish causation.
```

```python
#Is employee training associated with better productivity?
df["training_hours"].corr(df["productivity_score"])
```

```python
#Do employees with higher productivity scores tend to have fewer escalations?
df["productivity_score"].corr(df["escalations"])
```

```python
#Do longer resolution times correspond to lower customer satisfaction?
df["avg_resolution_time"].corr(df["customer_satisfaction"])
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    df["avg_resolution_time"],
    df["customer_satisfaction"],
    alpha=0.5
)

plt.xlabel("Average Resolution Time")
plt.ylabel("Customer Satisfaction")
plt.title("Resolution Time vs Customer Satisfaction")

plt.show()
```

```python
##An employee is High Risk if they have:
#5 or more escalations, AND
#2 or more recurring escalations
df["high_risk_employee"] = np.where(
    (df["escalations"] >= 5) &
    (df["recurring_escalations"] >= 2),
    "Yes",
    "No"
)
```

```python
df["high_risk_employee"].value_counts()
```

```python
#which departments have the most high-risk employees
high_risk_by_department = (
    df[df["high_risk_employee"] == "Yes"]
    .groupby("department")["employee_id"]
    .count()
    .sort_values(ascending=False)
)

high_risk_by_department
```

```python
high_risk_rate = (
    df.groupby("department")["high_risk_employee"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

high_risk_rate
```

```python
#What are the common escalation reasons among high-risk employees?
high_risk_reason = (
    df[df["high_risk_employee"] == "Yes"]
    ["escalation_reason"]
    .value_counts()
)

high_risk_reason
```

```python
#which teams have the most high-risk employees.
high_risk_by_team = (
    df[df["high_risk_employee"] == "Yes"]
    .groupby("team")["employee_id"]
    .count()
    .sort_values(ascending=False)
)

high_risk_by_team
```

```python
high_risk_team_rate = (
    df.groupby("team")["high_risk_employee"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

high_risk_team_rate
```

```python
#Do high-risk employees work more overtime?
df.groupby("high_risk_employee")["overtime_hours"].agg(
    ["mean", "median", "max"]
)
```

```python
#whether high-risk employees also have different productivity levels.
df.groupby("high_risk_employee")["productivity_score"].agg(
    ["mean", "median", "min", "max"]
)
```

```python
df.groupby("high_risk_employee")["avg_resolution_time"].agg(
    ["mean", "median", "min", "max"]
)
```

```python
#Do high-risk employees have lower customer satisfaction?
df.groupby("high_risk_employee")["customer_satisfaction"].agg(
    ["mean", "median", "min", "max"]
)
```

```python
#Which escalation reasons are associated with the longest resolution times?
reason_resolution = (
    df.groupby("escalation_reason")["avg_resolution_time"]
    .agg(["mean", "median", "count"])
    .sort_values("mean", ascending=False)
)

reason_resolution
```

```python
#which departments have the longest average resolution time.
department_resolution = (
    df.groupby("department")["avg_resolution_time"]
    .agg(["mean", "median", "count"])
    .sort_values("mean", ascending=False)
)

department_resolution
```

```python
correlation_cols = [
    "customer_interactions",
    "escalations",
    "recurring_escalations",
    "overtime_hours",
    "training_hours",
    "productivity_score",
    "avg_resolution_time",
    "customer_satisfaction"
]

correlation_matrix = df[correlation_cols].corr()

correlation_matrix
```

```python
#correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Workforce Metrics Correlation Matrix")
plt.show()
```

```python
training_productivity = (
    df.groupby(pd.cut(df["training_hours"], bins=4))
    ["productivity_score"]
    .agg(["mean", "count"])
)

training_productivity
```

```python
#whether training also relates to customer satisfaction
training_satisfaction = (
    df.groupby(pd.cut(df["training_hours"], bins=4))
    ["customer_satisfaction"]
    .agg(["mean", "count"])
)

training_satisfaction
```

```python
#High-risk rate by training level
training_risk = (
    df.groupby(pd.cut(df["training_hours"], bins=4))
    ["high_risk_employee"]
    .apply(lambda x: (x == "Yes").mean() * 100)
)

training_risk
```

```python
executive_summary = pd.DataFrame({
    "Metric": [
        "Total Employees",
        "Total Escalations",
        "Total Recurring Escalations",
        "Employee Escalation Rate (%)",
        "Recurring Escalation Rate (%)",
        "Average Resolution Time",
        "Average Customer Satisfaction",
        "High-Risk Employees",
        "High-Risk Employee Rate (%)"
    ],
    "Value": [
        len(df),
        df["escalations"].sum(),
        df["recurring_escalations"].sum(),
        (df["escalations"] > 0).mean() * 100,
        (df["recurring_escalations"] > 0).mean() * 100,
        df["avg_resolution_time"].mean(),
        df["customer_satisfaction"].mean(),
        (df["high_risk_employee"] == "Yes").sum(),
        (df["high_risk_employee"] == "Yes").mean() * 100
    ]
})

executive_summary
```
